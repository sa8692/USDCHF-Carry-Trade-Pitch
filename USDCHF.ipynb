```python
import yfinance as yf
import numpy as np
import pandas as pd
```

```python
data = yf.download('DX-Y.NYB', start='2010-01-01', end='2024-10-01')
data['Closing Vol'] = data['Close'].rolling(30).std()
```

```python
data
```

```python
new_data = {
    'Data': data.index.values,
    'Closing Vol': data['Closing Vol'].values
}

new_df = pd.DataFrame(new_data)
new_df = new_df[29:]
new_df
```

```python
np.mean(new_df['Closing Vol'])
```

```python
np.std(new_df['Closing Vol'])
```

```python
import matplotlib.pyplot as plt
plt.plot(new_df['Closing Vol'])
plt.show()
```
